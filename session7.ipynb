```python
# Q1

import matplotlib.pyplot as plt

instagram_hours = [1, 2, 2, 3, 4, 5, 6]
posts = [1, 1, 2, 3, 3, 4, 5]

plt.scatter(instagram_hours, posts)
plt.xlabel("Instagram Hours")
plt.ylabel("Posts")
plt.title("Instagram Hours vs Posts")
```

```python
# Q2

import numpy as np

zomato = [5, 8, 10, 12, 15, 18]
swiggy = [4, 7, 9, 11, 14, 17]

r = np.corrcoef(zomato, swiggy)

print(r)
print("Correlation =", r[0][1])
```
