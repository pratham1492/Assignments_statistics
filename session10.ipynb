```python
# Q1

def bayes_posterior(prior, likelihood, evidence):
    posterior = (prior * likelihood) / evidence
    return posterior

prior = 0.2
likelihood = 0.7
evidence = 0.4

print(bayes_posterior(prior, likelihood, evidence))
```

```python
# Q2

prior = 0.10
likelihood = 0.80

evidence = (0.80 * 0.10) + (0.20 * 0.90)

posterior = (prior * likelihood) / evidence

print("Probability restaurant is premium =", posterior)
```
