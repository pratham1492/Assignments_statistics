```python
# Q1

def calculate_probability(event_count, total_outcomes):
    return event_count / total_outcomes

heads = calculate_probability(1, 2)

print("Probability of getting Heads:", heads)
```

```python
# Q2

import random

count = 0

for i in range(100):
    dice = random.randint(1, 6)
    if dice == 6:
        count += 1

probability = count / 100

print("Number of 6s:", count)
print("Experimental Probability:", probability)
```

```python
# Q3

genres = [
    "pop","rock","pop","hiphop","pop",
    "jazz","rock","pop","classical","pop",
    "hiphop","pop","rock","pop","jazz",
    "pop","rock","hiphop","pop","pop"
]

pop_count = genres.count("pop")
probability = pop_count / len(genres)

print("Pop Songs:", pop_count)
print("Probability of selecting a Pop song:", probability)
```

```python
# Q4

def probability_both(food_users, dessert_users):
    food = set(food_users)
    dessert = set(dessert_users)

    both = food.intersection(dessert)
    total = food.union(dessert)

    return len(both) / len(total)

food_users = [1, 2, 3, 4, 5]
dessert_users = [3, 4, 5, 6]

print("Probability:", probability_both(food_users, dessert_users))
```
