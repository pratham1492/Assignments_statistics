```python
# Q3

from statistics import mode

def most_common_rating(ratings):
    return mode(ratings)

ratings = [5, 4, 4, 3, 5, 4, 2, 4]

print("Ratings:", ratings)
print("Mode =", most_common_rating(ratings))
```

```python
# Q4

import numpy as np
from statistics import mode

channel1 = [100, 120, 110, 105, 5000]
channel2 = [200, 210, 220, 230, 240]
channel3 = [150, 150, 160, 170, 180]

print("Channel 1")
print("Mean =", np.mean(channel1))
print("Median =", np.median(channel1))
print("Mode =", mode(channel1))

print()

print("Channel 2")
print("Mean =", np.mean(channel2))
print("Median =", np.median(channel2))
print("Mode =", mode(channel2))

print()

print("Channel 3")
print("Mean =", np.mean(channel3))
print("Median =", np.median(channel3))
print("Mode =", mode(channel3))

print()

print("Conclusion:")
print("Channel 1 is most affected by outliers because of the value 5000.")
```
