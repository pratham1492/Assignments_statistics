```python
# Q1

steps = [4500, 7000, 5000, 8000, 4000, 9000, 6000]

mean = sum(steps) / len(steps)

print("Mean =", mean)

total = 0

for i in steps:
    diff = i - mean
    square = diff ** 2
    total += square
    print(i, "-", round(mean, 2), "=", round(diff, 2), "Square =", round(square, 2))

variance = total / len(steps)

print("Sum of Squared Differences =", round(total, 2))
print("Variance =", round(variance, 2))
```

```python
# Q2

import math

def calculate_standard_deviation(scores):
    mean = sum(scores)/len(scores)

    variance = 0

    for i in scores:
        variance += (i - mean) ** 2

    variance = variance / len(scores)

    return round(math.sqrt(variance), 2)

scores = [75, 80, 90, 85, 70]

print(calculate_standard_deviation(scores))
```

```python
# Q3

import statistics

a = [200, 200, 200, 200, 200]
b = [100, 300, 150, 400, 50]

print("Friend A SD =", statistics.pstdev(a))
print("Friend B SD =", statistics.pstdev(b))
```

```python
# Q4

import math

salary = [28000, 29000, 27000, 60000, 26500, 27500]

mean = sum(salary) / len(salary)

total = 0

for i in salary:
    total = total + (i - mean) ** 2

variance = total / len(salary)

std = math.sqrt(variance)

print("Mean =", round(mean, 2))
print("Variance =", round(variance, 2))
print("Standard Deviation =", round(std, 2))
```
