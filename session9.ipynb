```python
# Q1

import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50,10,100)

plt.hist(data,bins=10)
plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
```

```python
# Q2

import numpy as np
from scipy import stats

steps = [6500,7000,7200,7500,7800,8000,8200,8500,8700,9000,9200,9400,9600,9800,10000,10200,10500,10800,11000]

mean = np.mean(steps)
median = np.median(steps)
mode = stats.mode(steps)

print("Mean =", mean)
print("Median =", median)
print("Mode =", mode.mode)
```
