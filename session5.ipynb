```python
# Q1

population = ["All Instagram Followers", "All Zomato Restaurants", "All Flipkart Products"]
sample = ["100 Followers", "50 Restaurants", "200 Products"]

print("Population:", population)
print("Sample:", sample)
```

```python
# Q2

import random

users = []

for i in range(1, 1001):
    users.append("user" + str(i))

sample = random.sample(users, 50)

for i in sample:
    print(i)
```

```python
# Q3

print("Sampling saves time and cost.")
print("A sample represents the whole population.")
```

```python
# Q5

random_sampling = ["Randomly select 500 users"]

stratified_sampling = [
    "18-25 -> Random users",
    "26-35 -> Random users",
    "36-50 -> Random users"
]

print("Random Sampling")
for i in random_sampling:
    print(i)

print("Stratified Sampling")
for i in stratified_sampling:
    print(i)
```
